```python
import sys
from pathlib import Path

BASE_DIR = Path.cwd()
if not (BASE_DIR / "data").exists():
    BASE_DIR = BASE_DIR.parent
sys.path.insert(0, str(BASE_DIR))

import pandas as pd
from src.preprocessing import load_dataset, prepare_features, split_dataset, split_features_target

DATA_PATH = BASE_DIR / "data" / "WA_Fn-UseC_-Telco-Customer-Churn.csv"
Telco_cust_churn_data = load_dataset(DATA_PATH)
print(f"Loaded dataset: {Telco_cust_churn_data.shape[0]} rows, {Telco_cust_churn_data.shape[1]} columns")
```

```python
X = prepare_features(Telco_cust_churn_data)
print("Feature columns after removing identifiers and target:")
print(X.columns.tolist())
```

```python
X, y = split_features_target(Telco_cust_churn_data)

numeric_columns = X.select_dtypes(include=["number"]).columns.tolist()
categorical_columns = X.select_dtypes(exclude=["number"]).columns.tolist()

print("Numerical columns:", numeric_columns)
print("Categorical columns:", categorical_columns)
print("Encoded target labels:", sorted(y.unique().tolist()))
```

```python
X_train, X_test, y_train, y_test = split_dataset(Telco_cust_churn_data)

print("Training set shape:", X_train.shape)
print("Test set shape:", X_test.shape)
print("Train class distribution:", y_train.value_counts().sort_index().to_dict())
print("Test class distribution:", y_test.value_counts().sort_index().to_dict())
```
