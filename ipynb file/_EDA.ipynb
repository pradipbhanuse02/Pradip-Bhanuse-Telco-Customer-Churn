```python


# Create output directory for plots
output_dir = BASE_DIR / "plots"
output_dir.mkdir(exist_ok=True)

# Find numerical columns
numerical_cols = Telco_cust_churn_data.select_dtypes(include=['int64', 'float64']).columns.tolist()
print(f"Numerical columns: {numerical_cols}")

# Create histplots for each numerical column
for col in numerical_cols:
    plt.figure(figsize=(10, 5))
    sns.histplot(data=Telco_cust_churn_data, x=col, kde=True, bins=30)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    
    # Save the plot
    plot_path = output_dir / f"{col}_histplot.png"
    plt.savefig(plot_path, dpi=300, bbox_inches='tight')
    print(f"Saved: {plot_path}")
    plt.close()

print(f"\nAll histplots saved to: {output_dir}")
```

```python
# Find non-numerical (categorical) columns
categorical_cols = Telco_cust_churn_data.select_dtypes(include=['object']).columns.tolist()

# Filter out ID columns and columns with too many unique values
columns_to_plot = [col for col in categorical_cols 
                   if col not in ['customerID'] 
                   and Telco_cust_churn_data[col].nunique() < 20]

print(f"Categorical columns to visualize: {columns_to_plot}\n")

# Create countplots for each categorical column
for col in columns_to_plot:
    num_categories = Telco_cust_churn_data[col].nunique()
    fig_width = max(8, num_categories * 0.6)
    
    plt.figure(figsize=(fig_width, 5))
    sns.countplot(data=Telco_cust_churn_data, x=col, order=Telco_cust_churn_data[col].value_counts().index)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    
    # Save the plot
    plot_path = output_dir / f"{col}_countplot.png"
    plt.savefig(plot_path, dpi=100, bbox_inches='tight')
    print(f"Saved: {plot_path}")
    plt.close()

print(f"\nAll categorical countplots saved to: {output_dir}")
```
